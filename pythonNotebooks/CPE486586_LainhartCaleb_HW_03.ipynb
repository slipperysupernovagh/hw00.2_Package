##7) Computing Derivative using PyTorch Autograd

```python
import torch
rs = torch.tensor(0.00887, requires_grad = True)
r = torch.tensor(30000.1, requires_grad = True)
tau = 2*r*torch.sqrt(r/rs)*torch.sqrt(1-(rs/r))
tau.backward()
print(f"dtau/dr: {r.grad}")
```

##8) Linear Regression in One Variable with Scikit-Learn

```python
import polars as pl
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.model_selection import train_test_split, cross_val_score
from statistics import mean
import seaborn as sns
import matplotlib.pyplot as plt

data = pl.read_csv("https://raw.githubusercontent.com/rahulbhadani/Datasets/refs/heads/master/RAV4/2020-04-15-05-12-00_2T3Y1RFV8KC014025_CAN_Messages_5_Hz.csv")
x = data[['steer_angle']]
y = data['yaw_rate']

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size = 0.25)
linearModel = LinearRegression()
linearModel.fit(X_train, y_train)
print(f"Linear model coefficients: y= {linearModel.coef_}x + {linearModel.intercept_}")
print(f"R^2: {linearModel.score(X_test,y_test)}")

sns.scatterplot(x='steer_angle', y='yaw_rate', data=data)
plt.xlabel("Steer Rate")
plt.ylabel("Yaw Rate")
yhat = linearModel.coef_*x + linearModel.intercept_
plt.plot(x,yhat,color='orange')
```
